## 6. Calculate the factorial of a number.

```python
num = int(input("Enter the number: "))
factorial = 1
count = 0
if num < 0:
    print("There is no factorial for negetive numbers")
else:    
    while count < num:
        factorial = factorial * (num - count)
        count +=1
    print(f"The factorial of {num} is {factorial}")
```

## 16. Check whether a number is an Armstrong number.

```python
num = int(input("Enter the number"))
act_num = num
length = len(str(num))
total = 0
while num // 10 != 0:
    total = total + ((num % 10) ** length)
    num = num // 10
total = total + (num ** length)
if total == act_num:
    print("It is an Armstrong number")
else:
    print("It is not an Armstrong number")
```

## 32. Check whether a number is a strong number.

```python
num = input("Enter the number")
temp = num
result = 0
loop = len(num)
if int(num) > 0:
    while loop >= 1:
        count = int(num[loop -1])
        fact = 1
        if count > 0:
            while count >=1:
                fact = fact * count
                count -= 1
        result = result + fact
        loop -= 1
    if str(result) == temp:
        print(f"{temp} is a strong number")
    else:
        print(f"{temp} is not a strong number")
else:
       print("Please enter a positive number")
```
